```python
import pandas as pd

# Cargar datos
ruta_archivo = 'C:/Users/edamv/retail-sales-analyisis/retail_sales_dataset.csv'
df = pd.read_csv(ruta_archivo)

# Mostrar las primeras 10 filas
print(df.head(10))
```

```python
#Exploración Inicial de los Datos 

# Últimas 5 filas
print(df.tail(5))

# Información general
print(df.info())

# Estadísticas descriptivas
print(df.describe())
```

```python
#Inspección de Datos
# Tipos de datos
print(df.dtypes)

# Valores únicos en la columna 'Product Category'
print(df['Product Category'].value_counts())

# Valores únicos en la columna 'Tienda'
print(df['Transaction ID'].unique())
```

```python
# Filtrado de Datos
# Ventas mayores a 50
ventas_mayores_50 = df[df['Total Amount'] > 50]
print(ventas_mayores_50)

# Price per Unit menor a 0.5
precio_menor_0_5 = df[df['Price per Unit'] < 0.5]
print(precio_menor_0_5)

# Product Category sea 'Electronics' y 'Total Amount' mayores a 30

electronics_mayores_30 = df.query('`Product Category` == "Electronics" and `Total Amount` > 30')
print(electronics_mayores_30)

```

```python
#Slicing de Datos
# Seleccionar solo las columnas 'Product Category' y 'Total Amount'
print(df[['Product Category', 'Total Amount']])
```

```python
# Utilizando loc[], filas de la 5 a la 10 y columnas 'Product Category' y 'Price Unit'
print(df.loc[5:10, ['Product Category', 'Price per Unit']])
```

```python
# Utilizando iloc[], primeras 5 filas y 3 columnas
print(df.iloc[:5, :3])
```
