```python
import numpy as np
```

```python
#Carga de datos
def cargar_datos(ruta_archivo):
    datos = np.genfromtxt(ruta_archivo, delimiter=',', skip_header=1, dtype=None, encoding=None)
    return datos

def eliminar_nulos(datos):
    return datos[~np.isnan(datos[:, -1].astype(float))]

def eliminar_duplicados(datos):
    # Convierte cada fila a una tupla y usa un conjunto para eliminar duplicados
    datos_unicos = list(set(tuple(row) for row in datos))
    # Convierte de nuevo a un array numpy
    return np.array(datos_unicos)

ruta_archivo = 'C:\\Users\\edamv\\retail-sales-analyisis\\retail_sales_dataset.csv'
datos = cargar_datos(ruta_archivo)
datos = eliminar_duplicados(datos)
datos = eliminar_nulos(datos)

print(datos)
print(datos[:, -4]) #Product Category
print(datos[:, -1]) #Total Amount
```

```python
#Exploración de datos
# Suponiendo que las columnas están en el orden ['Product Category', ..., 'Date', ..., 'Total Amount']
categorias = datos[:, -4]  # 'Product Category'
ventas = datos[:, -1].astype(float)  # 'Total Amount'

def calcular_totales_ventas_por_categoria(categorias, ventas):
    resultado = np.unique(categorias, return_inverse=True)
    unicos = resultado[0]
    indices = resultado[1]
    totales = np.array([ventas[indices == i].sum() for i in range(len(unicos))])
    return unicos, totales

def calcular_promedio_ventas_diarias_por_categoria(categorias, ventas):
    resultado = np.unique(categorias, return_inverse=True)
    unicos = resultado[0]
    indices = resultado[1]
    promedios = np.array([ventas[indices == i].mean() for i in range(len(unicos))])
    return unicos, promedios

def encontrar_mayores_menores_ventas(totales):
    if len(totales) == 0:
        print("Error: Secuencia vacía pasada a `argmax`.")
        return None, None
    print(f"Totales recibidos: {totales}")
    mayor_ventas = np.argmax(totales, axis=0)
    menor_ventas = np.argmin(totales, axis=0)
    return mayor_ventas, menor_ventas

```

```python
# Calcular Totales y Promedios
resultado = calcular_totales_ventas_por_categoria(categorias, ventas)
categorias_unicas = resultado[0]
totales_por_categoria = resultado[1]

resultado = calcular_promedio_ventas_diarias_por_categoria(categorias, ventas)
promedios_por_categoria = resultado[1]

resultado = encontrar_mayores_menores_ventas(totales_por_categoria)
categoria_mayor_ventas = resultado[0]
categoria_menor_ventas = resultado[1]
```

```python
#Imprimir resultados

if categoria_mayor_ventas is not None and categoria_menor_ventas is not None:
    print("Totales de ventas por categoría:", totales_por_categoria)
    print("Promedios de ventas diarias por categoría:", promedios_por_categoria)
    print(f"Categoría con mayor ventas: {categorias_unicas[categoria_mayor_ventas]}")
    print(f"Categoría con menor ventas: {categorias_unicas[categoria_menor_ventas]}")
else:
    print("No se encontraron categorías válidas para calcular ventas.")
```
