```python
import pandas as pd

# Cargar datos
ruta_archivo = 'C:/Users/edamv/retail-sales-analyisis/retail_sales_dataset.csv'
df = pd.read_csv(ruta_archivo)

# Mostrar las primeras 10 filas
print(df.head(10))

```

```python
#Transformación de Datos

#Nueva columna
df['Total_Income'] = df['Quantity'] * df['Price per Unit']
print(f"Total Income: {df['Total_Income']}")

#Normalización de columna
df['Normal_Total_Amount'] = (df['Total Amount'] - df['Total Amount'].mean()) / df['Total Amount'].std()
print(f"Normal_Total_Amount: {df['Normal_Total_Amount']}")
```

```python
#Clasificación de Datos
df['Total_Amount_Category'] = pd.cut(df['Total Amount'], bins=[0, 100, 500, 1000], labels=['Low', 'Medium', 'High'])
print(f"Total Amount: {df['Total Amount']} , Total_Amount_Category: {df['Total_Amount_Category']}")

```

```python
#Agrupación 
group_product_price = df.groupby(['Product Category', 'Price per Unit']).agg({
    'Total Amount': ['sum', 'mean', 'count', 'min', 'max', 'std', 'var'],
    'Total_Income': ['sum', 'mean', 'count', 'min', 'max', 'std', 'var']
})
print(group_product_price)
```

```python
#Análisis con apply
def standard_amount(media):
    return lambda x: x - media

df['Standard_Amount'] = df.groupby('Product Category')['Total Amount'].transform(standard_amount(df['Total Amount'].mean()))
print(df[['Product Category', 'Total Amount', 'Standard_Amount']])
```
